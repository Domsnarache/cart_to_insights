#### Data Quality Audit 
---
1. Missing value distribution
2. Exact duplicate records
3. Data type inconsistencies (string timestamps, integer postal codes)

### Data Quality Audit Findings

1. **Duplicate Records:**
   - `df_geolocation` contains 261,831 exact duplicate rows. Zero duplicates found in the other 8 tables.

2. **Data Type Remediations:**
   - 8 timestamp columns are stored as text (`object`) and require conversion to `datetime64[ns]`:
     * `df_orders`: purchase, approved, carrier delivery, customer delivery, estimated delivery.
     * `df_order_items`: shipping_limit_date.
     * `df_reviews`: creation_date, answer_timestamp.
   - Postal code prefix fields across customers, sellers, and geolocation are stored as integers (`int64`). Cast to 5-digit zero-padded strings (`str.zfill(5)`) to preserve leading zeros.

3. **Domain Validation:**
   - `review_score`: Valid range 1 to 5 verified.
   - `payment_type`: 3 rows contain `'not_defined'`; reclassify to `'other'`.
   - `payment_value`: 9 transactions with `0.00` payment (100% discount vouchers). Zero negative amounts.
   - `product_weight_g`: 4 items have non-positive weight (`<= 0`); impute with category median.
   - `price`: Zero non-positive prices found.

4. **Outlier Boundaries:**
   - `price`: Median ~75 BRL vs maximum 6,735 BRL (high-ticket items).
   - `freight_value`: Median ~16 BRL vs maximum 409 BRL (interstate heavy cargo).
   - `product_weight_g`: Median 700g vs maximum 40.4 kg.
   - `geolocation`: Extreme GPS coordinates detected outside Brazilian territory (positive latitude +45°, longitude +121°); clip to national geographic boundaries.

5. **Key Cardinality:**
   - Primary Keys (`order_id`, `customer_id`, `product_id`, `seller_id`, `product_category_name`) verified 100% unique.
   - Foreign Keys demonstrate expected 1-to-many cardinality in `order_items` and `order_payments`.
