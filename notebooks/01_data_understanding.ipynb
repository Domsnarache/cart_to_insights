```python
import pandas as pd
from pathlib import Path
```

```python
# Discover raw CSV datasets
RAW_DATA_DIR: Path = Path("../data/raw")

csv_files: list[Path] = []
for file in RAW_DATA_DIR.glob("*.csv"):
    csv_files.append(file)

csv_files.sort()

print("Total CSV files found :", len(csv_files))
for file_name in csv_files:
    print(file_name.name)
```

```python
# Load all 9 raw datasets into DataFrames
df_orders: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_orders_dataset.csv") 
df_customers: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_customers_dataset.csv")
df_order_items: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_order_items_dataset.csv")
df_payments: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_order_payments_dataset.csv")
df_reviews: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_order_reviews_dataset.csv")
df_products: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_products_dataset.csv")
df_sellers: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_sellers_dataset.csv")
df_geolocation: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "olist_geolocation_dataset.csv")
df_category_translation: pd.DataFrame = pd.read_csv(RAW_DATA_DIR / "product_category_name_translation.csv")

# Verify row and column dimensions across tables
print("All 9 Tables Loaded and its Rows/Columns:")
print("-" * 45)
print("1. Orders              :", df_orders.shape)
print("2. Customers           :", df_customers.shape)
print("3. Order Items         :", df_order_items.shape)
print("4. Order Payments      :", df_payments.shape)
print("5. Order Reviews       :", df_reviews.shape)
print("6. Products            :", df_products.shape)
print("7. Sellers             :", df_sellers.shape)
print("8. Geolocation         :", df_geolocation.shape)
print("9. Category Translation:", df_category_translation.shape)
```

```python
# Profile schema, column types, and null counts across tables
tables: dict[str, pd.DataFrame] = {
    "Orders": df_orders,
    "Customers": df_customers,
    "Order Items": df_order_items,
    "Order Payments": df_payments,
    "Order Reviews": df_reviews,
    "Products": df_products,
    "Sellers": df_sellers,
    "Geolocation": df_geolocation,
    "Category Translation": df_category_translation
}

for table_name, df in tables.items():
    print(f"Table:{table_name}")
    df.info()
```

```python
# Analyze customer cardinality: order transaction IDs vs unique human buyers
total_transaction_ids: int = df_customers['customer_id'].nunique()
actual_unique_customers: int = df_customers['customer_unique_id'].nunique()
repeat_orders_count: int = total_transaction_ids - actual_unique_customers

print("Total Transaction IDs (Customer_id):", total_transaction_ids)
print("Actual Unique Customers (Customer_unique_id):", actual_unique_customers)
print("Repeated Customer IDs (Difference) :", repeat_orders_count)
```
